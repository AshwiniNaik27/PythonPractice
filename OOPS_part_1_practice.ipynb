```python
# Week 1 Student Lab: Object-Oriented Programming (OOP) Part 1
##**Objective:** Practice creating Classes, Objects, Instance Variables, `__init__`, and Instance Methods from scratch.
## Exercise 1: Build a `Car` Class

# Create a class named Car that:
# 1. Takes brand, model, and price as parameters in __init__.
# 2. Has an instance variable speed initialized to 0.
# 3. Has a method accelerate(amount) that increases speed by amount and prints the current speed.
# 4. Has a method display_info() that prints brand, model, price, and current speed.

class Car:

  def __init__(self, brand: str, model:str, price:float):
          self.brand = brand
          self.model = model
          self.price = price
          self.speed = 0

  def accelerate(self, amount: int):
    self.speed = self.speed + amount
    print(self.speed)

  def display_info(self):
    print("Brand: ", self.brand)
    print("Model: ", self.model)
    print("Price: ", self.price)
    print("Current Speed: ", self.speed)

```

```python
car1 = Car("Toyota", "Innova" ,2500000 )
car1.display_info()
```

```python
car1.accelerate(30)
car1.accelerate(20)
```

```python
car1.display_info()
```

```python
My_Car = Car("Tata", "Nexon", 1200000)
My_Car.display_info()

My_Car.accelerate(40)
My_Car.accelerate(20)

My_Car.display_info()
```

```python
## Exercise 2: Build a BankAccount System
# Create a class BankAccount that:
# 1. Takes account_holder and initial balance in __init__.
# 2. Has a deposit(amount) method that adds amount to balance.
# 3. Has a withdraw(amount) method that subtracts amount from balance (if sufficient funds exist).
# 4. Has a get_balance() method that returns the current balance.


class BankAccount:
  def __init__(self, account_holder : str, initial_balance : float):
         self.account_holder = account_holder
         self.initial_balance = initial_balance

  def deposit(self, amount:float):
     self.initial_balance += amount
     print(amount, "deposited sucessfully")

  def withdraw(self, amount: float):
    if amount <= self.initial_balance:
        self.initial_balance -= amount
        print(amount, "withdrawn sucessfully")
    else:
      print("insufficiant balance")

  def get_balance(self):
     return self.initial_balance
```

```python
acc = BankAccount("Priya", 5000)

acc.deposit(2500)
acc.withdraw(1000)
acc.get_balance()
```
