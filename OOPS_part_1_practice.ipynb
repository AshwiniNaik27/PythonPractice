```python
# Week 1 Student Lab: Object-Oriented Programming (OOP) Part 1
##**Objective:** Practice creating Classes, Objects, Instance Variables, `__init__`, and Instance Methods from scratch.
## Exercise 1: Build a `Car` Class

# Create a class named Car that:
# 1. Takes brand, model, and price as parameters in __init__.
# 2. Has an instance variable speed initialized to 0.
# 3. Has a method accelerate(amount) that increases speed by amount and prints the current speed.
# 4. Has a method display_info() that prints brand, model, price, and current speed.

class Car:

  def __init__(self, brand: str, model:str, price:float):
          self.brand = brand
          self.model = model
          self.price = price
          self.speed = 0

  def accelerate(self, amount: int):
    self.speed = self.speed + amount
    print(self.speed)

  def display_info(self):
    print("Brand: ", self.brand)
    print("Model: ", self.model)
    print("Price: ", self.price)
    print("Current Speed: ", self.speed)

```

```python
car1 = Car("Toyota", "Innova" ,2500000 )
car1.display_info()
```

```python
car1.accelerate(30)
car1.accelerate(20)
```
