```python
# Print even and odd numbers up to N

N = int(input("enter the number: "))

for i in range(1, N+1):
  if i % 2 == 0:
   print(i, "is even")
  else:
    print(i," is odd")
```

```python
# Sum of the first N natural numbers

N = int(input("Enter the number: "))
sum = 0

for i in range(1, N+1):
  sum = sum + i
print(sum)
```

```python
# Factorial of a number
# 4! = 4*3*2*1

N = int(input("Enter the number: "))
fact = 1

for i in range(1, N + 1):
    fact = fact * i

print("Factorial =", fact)
```

```python
# Reverse a number

N = int(input("Enter the number: "))
rev = 0
while N > 0:
  digit = N % 10
  rev = rev * 10 + digit
  N = N // 10
print(rev)
```

```python
# Sum of the digits of a number

N = int(input("Enter the number: "))
sum = 0

while N > 0:
  digit = N % 10
  sum = sum + digit
  N = N // 10
print(sum)
```

```python
# Check whether a number is a palindrome

N = int(input("Enter the number: "))
original = N
rev = 0

while N > 0:
  digit = N % 10
  rev = rev * 10 + digit
  N = N // 10

if original == rev:
  print("number is a palindrome")
else:
  print("number is not palindrome")


```
