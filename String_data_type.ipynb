```python
#Create Short Form from initial character
#Given a string create short form ofthe string from Initial character. Short form should be capitalised.
# input- Data science mentorship program
# output -DSMP



s = "Data science mentorship program"
short_form = ""

for word in s.split():
  short_form = short_form + word[0]
print(short_form.upper())
```

```python
#Append second string in the middle of first string
#First string:HelloWorld
#Second string:Python

s1 = "HelloWorld"
s2 = "Python"

mid = len(s1) // 2
result = s1[:mid] + s2 + s1[mid:]
print(result)
```

```python
#Given string contains a combination of the lower and upper case letters.
#Write a program to arrange the characters of a string so that all lowercase letters should come first.
# str1 = PyNaTive
# Expected Output: yaivePNT


str1 = "PyNaTive"

lower = ""
upper = ""

for i in str1:
  if i.islower() :
    lower = lower + i
  else:
    upper = upper + i
print (lower + upper)
```

```python
#Take a alphanumeric string input and print the sum and average of the digits that appear in the string, ignoring all other characters.
#Input:hel123O4every093


str = "hel123O4every093"

sum = 0
count = 0

for i in str:
  if i.isdigit():
    sum = sum + int(i)
    count = count + 1

avg = sum / count

print("Sum: ", sum)
print("Avg: ", round(avg,2))

```

```python
#Given a string. the task is to check if the string is symmetrical or not.
#A string is said to be symmetrical if both the halves of the string are the same.
#input=khokho

str = input("enter the string: ")

mid = len(str) // 2

if str[:mid]==str[mid:]:
  print("string is symmetrical")
else:
  print("string is not symmetrical")
```
