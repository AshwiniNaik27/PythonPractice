```python
# before pydantic

class Student:
    def __init__(self, name: str, age: int, gpa: float):
        self.name = name
        self.age = age
        self.gpa = gpa


s = Student(name=99, age="Twenty", gpa=False)

print(s.name, "", s.age, "", s.gpa)
```

```python
# Pydantic

from pydantic import BaseModel

class Student(BaseModel):
    name: str
    age: int
    gpa: float

# sending Correct data
s1 = Student(name="Ram", age=21, gpa=2.9)

print(s1)

# sending Incorrect data
s2 = Student(name=99, age="Twenty", gpa="excellent")

print(s2)
```

```python
#Real-world API dictionary

raw_data = {
    "name": "Sham",
    "age": 20,
    "gpa": 3.5
}

s2 = Student(**raw_data)

print(s2)
```

```python
raw_data1 = {
    "name": "Sham",
    "age": 21,
    "gpa": 2.15
}

s3 = Student(**raw_data1)
print(s3)
```

```python
# ValidationError

from pydantic import ValidationError

raw_data4 = {
    "name": 1234,
    "age": "Twenty"
}

s4 = Student(**raw_data4)
```

```python
# how to handle ValidationError

from pydantic import ValidationError

raw_data4 = {
    "name": 1234,
    "age": "Twenty"
}

try:
    s4 = Student(**raw_data4)
except ValidationError as e:
    print(f"Total number of errors found: {e.error_count()}")

    for err in e.errors():
        print(f"Fields : {err['loc']} , Type : {err['type']} , Message: {err['msg']}")
```

```python
# H.W

from pydantic import BaseModel, ValidationError


class Product(BaseModel):
    product_id: int
    name: str
    price: float
    in_stock: bool

raw_data1 = {
    "product_id": "50",
    "name": "Laptop",
    "price": "199.99",
    "in_stock": "1"
}

p1 = Product(**raw_data1)

print(p1)
print("Product ID:", p1.product_id)
print("Name:", p1.name)
print("Price:", p1.price)
print("In Stock:", p1.in_stock)



raw_data2 = {
    "product_id": "abc",
    "name": 1234,
    "price": "not available",
    "in_stock": "maybe"
}

try:
    p2 = Product(**raw_data2)

except ValidationError as e:

    print(f"Total number of errors found: {e.error_count()}")

    for err in e.errors():
        print(f"Fields : {err['loc']} , " f"Type : {err['type']} , " f"Message: {err['msg']}")
```

```python
# Required and nullable
# what happens when value is not provided in dict

from pydantic import BaseModel, ValidationError

class Product(BaseModel):
    product_id: int
    name: str | None

value = {
          "product_id": "50"
}

try:
    p1 = Product(**value)
except ValidationError as e:
    print(f"Total number of error: {e.error_count()}")

    for err in e.errors():
        print(f"Fields : {err['loc']} , Type : {err['type']} , Message: {err['msg']}")
```

```python
# with None

from pydantic import BaseModel, ValidationError

class Product(BaseModel):
    product_id: int
    name: str | None

value = {
    "product_id": "50",
    "name": None
}

try:
    p1 = Product(**value)
    print(p1)

except ValidationError as e:
    print(f"Total number of error: {e.error_count()}")

    for err in e.errors():
        print(f"Fields : {err['loc']} , "f"Type : {err['type']} , "f"Message: {err['msg']}")
```
