```python
# Check whether a triangle is equilateral, isosceles, or scalene

a = int(input("enter first side of traingle: "))
b = int(input("enter second side of traingle: "))
c = int(input("enter third side of traingle: "))

if a == b == c:
  print("triangle is equilateral")
elif a == b or b == c or c == a:
  print("triangle is isosceles")
else:
  print("triangle is scalene")
```

```python
# Simple calculator (+, -, *, /) using if-elif-else


num1 = int(input("enter the first number:"))
num2 = int(input("enter the second number:"))

operator = input("enter the operator: ")

if operator == "+":
  print(num1 + num2)
elif operator == "-":
  print(abs(num1 - num2))
elif operator == "*":
  print(num1 * num2)
elif operator == "/":
  print(num1 / num2)
else:
  print("pls enter correct mathametical operator")
```

```python
# Calculate an electricity bill with slab rates

units = int(input("enter the electricity unit:"))
if units <= 100:
  bill = units * 5
elif units <= 200:
  bill = (100 * 5) + (units - 100) * 7
else:
  bill = (100 * 5) + (100 * 7) + (units - 200) * 10
print("Electricity bill is:" , bill)
```
