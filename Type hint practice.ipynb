```python
# 1. BASIC TYPE HINTS

name: str = "Ashwini"
age: int = 30
salary: float = 12.5
is_working: bool = True

print(name)
print(age)
print(salary)
print(is_working)
```

```python
# 2. TYPE HINT IN FUNCTION

def add(a: int, b: int) -> int:
    return a + b


result = add(10, 20)
print("Addition:", result)
```

```python
# 3. STRING TYPE HINT

def greet(name: str) -> str:
    return "Hello " + name


print(greet("Ashwini"))
```

```python
# 4. FLOAT TYPE HINT

def calculate_salary(basic: float, bonus: float) -> float:
    return basic + bonus


total_salary = calculate_salary(50000.0, 5000.0)
print("Total Salary:", total_salary)
```

```python
# 5. BOOLEAN TYPE HINT

def check_status(is_active: bool) -> str:
    if is_active:
        return "Employee is active"
    else:
        return "Employee is inactive"


print(check_status(True))
print(check_status(False))
```

```python
# 6. LIST TYPE HINT

def calculate_total(numbers: list[int]) -> int:
    return sum(numbers)


numbers = [10, 20, 30, 40]

print("Total:", calculate_total(numbers))
```

```python
# 7. LIST OF STRINGS

def print_names(names: list[str]) -> None:
    for name in names:
        print(name)


employees = ["Ashwini", "Rahul", "Priya"]

print_names(employees)
```

```python
# 8. DICTIONARY TYPE HINT

def get_salary(employee: dict[str, int]) -> int:
    return employee["salary"]


employee = {
    "salary": 50000
}

print("Salary:", get_salary(employee))
```

```python
# 9. DICTIONARY WITH STRING VALUES

def get_employee_name(employee: dict[str, str]) -> str:
    return employee["name"]


employee = {
    "name": "Ashwini",
    "department": "NOC"
}

print("Employee:", get_employee_name(employee))
```

```python
# 10. TUPLE TYPE HINT

def get_employee_details() -> tuple[str, int]:
    return "Ashwini", 30


name, age = get_employee_details()

print("Name:", name)
print("Age:", age)
```

```python
# 11. SET TYPE HINT

def get_unique_numbers(numbers: set[int]) -> set[int]:
    return set(numbers)


numbers = {10, 20, 20, 30, 30}

print("Unique Numbers:", get_unique_numbers(numbers))
```

```python
# 12. OPTIONAL TYPE HINT

from typing import Optional


def find_employee(employee_id: int) -> Optional[str]:

    if employee_id == 101:
        return "Ashwini"

    return None


print(find_employee(101))
print(find_employee(999))
```

```python
# 13. UNION TYPE HINT

def display_id(employee_id: int | str) -> None:
    print("Employee ID:", employee_id)


display_id(101)
display_id("EMP101")
```

```python
# 14. ANY TYPE

from typing import Any


def display_data(data: Any) -> None:
    print(data)


display_data(100)
display_data("Ashwini")
display_data([10, 20, 30])
display_data({"name": "Ashwini"})
```

```python
# 15. FUNCTION RETURNING NONE

def welcome(name: str) -> None:
    print("Welcome", name)


welcome("Ashwini")
```
