```python
# 4.Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:
# Salary(Lakhs) : Tax(%)

# Below 5 : 0%
# 5-10 : 10%
# 10-20 : 20%
# aboove 20 : 30%

CTC=float(input("enter anual CTC in lakh: "))

HRA= CTC * 0.10
DA= CTC * 0.05
PF= CTC * 0.03

if CTC < 5:
    tax = CTC * 0
elif CTC <= 10:
    tax = CTC * 0.10
elif CTC <= 20:
    tax = CTC * 0.20
else:
    tax = CTC * 0.30

inhand = (CTC - HRA - DA - PF - tax) / 12

print("In-hand monthly salary:", round(inhand, 2), "Lakhs")
print("In-hand salary in rupees:", round(inhand * 100000,2))

```

```python
#8.Find the factorial of a given number.

n=int(input("Enter the number: "))

fac=1
i=1 #counter

if i<0 :
  print("factorial is not possible for negative number")
else:
  while i<=n:
    fac = fac * i
    i = i + 1
  print ("Factorial = " ,fac)

```

```python
#9. Reverse a given integer number.

n = int(input("Enter the number: "))
reverse=0

while n > 0 :
  digit = n % 10
  reverse = reverse * 10 + digit
  n = n // 10
print ("Reverse= ", reverse )

```

```python
#10. ake a user input as integer N.
# Find out the sum from 1 to N. If any number if divisible by 5, then skip that number.
# And if the sum is greater than 300, don't need to calculate the sum further more.
# Print the final result.
N = int (input("Enter the number: "))

i = 1
sum = 0
while i <= N :
  if i % 5 == 0 :
      i = i + 1
      continue
  elif sum + i > 300 :
     break
  else:
   sum = sum + i
   i = i + 1

print ("Sum :", sum)
```

```python
#12. Write a program which will find all such numbers which are divisible by 7 but are not a multiple of 5, between 2000 and 3200 (both included).
#The numbers obtained should be printed in a comma-separated sequence on a single line.


i = 2000

while i <= 3200:
    if i % 7 == 0 and i % 5 != 0:
        print(i, end=",")
    i = i + 1

```

```python
#14. #13. Write a program, which will find all such numbers between 1000 and 3000 (both included) such that each digit of the number is an even number.
#The numbers obtained should be printed in a space-separated sequence on a single line.

for i in range(1000, 3000) :
  if all(int(digit) % 2 == 0 for digit in str(i)):
    print(i, end=" ")

```

```python
#16.Write a program to print whether a given number is a prime number or not

n = int (input ("Enter the number: "))

if n <= 1:
  print ("not a prime number")
else:
  for i in range(2, n):
      if n % i == 0:
         print ("not a prime number")
         break
  else:
      print ("prime number")
```

```python
#17. Print all the Armstrong numbers in a given range.


n = int (input("enter the number: "))
sum = 0
for digit in str(n):
  sum = sum + int(digit) ** 3
if sum == n:
    print ("It is an Armstrong number.")
else:
    print ("not an Armstrong number.")

```
