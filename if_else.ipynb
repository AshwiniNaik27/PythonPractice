```python
#Write a Python program to check whether the entered email and password are correct.
#If both are correct, print "Login successful".
#Otherwise, ask the user to enter the email and password one more time.


#first need to give input

email=input("Enter the email: ")
password=input("Enter the password: ")

if email=='ashwini.nemlekar@gamilcom' and password=='1234':
  print ("login successful")
elif email=='ashwini.nemlekar@gamilcom' and password!='1234':
  print ('incorrect password')
  password=input("Enter password again: ")
  if password=='1234':
   print ('login sucessful')
  else:
   print('Pls try after 24hrs')
elif email!='ashwini.nemlekar@gamilcom' and password=='1234':
  print ('incorrect email')
  email=input('Enter email again')
  if email=='ashwini.nemlekar@gamilcom':
    print('login sucessful')
  else:
    print('Pls try after 24hrs')
else:
   print('email or password is incorrect')
```
