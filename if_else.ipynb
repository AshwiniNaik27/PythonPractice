```python
# Write a Python program to accept two numbers and a mathematical operator from the user.
# If the user enters any other operator or leaves it blank, print "Please use correct mathematical operation".


fnum=int(input("enter fnum: "))
snum=int(input("enter snum: "))

operator=input("enter operator: ")

if operator=="+" :
  print(fnum+snum)
elif operator=="-" :
  print(fnum-snum)
elif operator=="*" :
  print(fnum*snum)
elif operator=="/":
  print(fnum/snum)
else:
  print("Please use correct mathematical operation")

```

```python
# Write a Python program to accept three numbers from the user.
# 1.Check which number is the minimum among the three numbers.
# 2.If all three numbers are the same, print "All three numbers are equal"


#1st i have to give input

a=int(input('Enter first number:'))
b=int(input('Enter second number:'))
c=int(input('Enter third number:'))

if a==b==c :
  print('All three numbers are equal')
elif a<b and a<c :
  print('minimum number is :',a)
elif b<c :
  print('minimum number is: ',b)
else:
  print('minimum number is: ',c)
```

```python
# Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:
# Salary(Lakhs) : Tax(%)

# Below 5 : 0%
# 5-10 : 10%
# 10-20 : 20%
# aboove 20 : 30%

CTC=float(input("enter anual CTC in lakh: "))

HRA= CTC * 0.10
DA= CTC * 0.05
PF= CTC * 0.03

if CTC < 5:
    tax = CTC * 0
elif CTC <= 10:
    tax = CTC * 0.10
elif CTC <= 20:
    tax = CTC * 0.20
else:
    tax = CTC * 0.30

inhand = (CTC - HRA - DA - PF - tax) / 12

print("In-hand monthly salary:", round(inhand, 2), "Lakhs")
print("In-hand salary in rupees:", round(inhand * 100000,2))

```
