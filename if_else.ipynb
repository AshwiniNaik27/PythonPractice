```python
#Write a Python program to check whether the entered email and password are correct.
#If both are correct, print "Login successful".
#Otherwise, ask the user to enter the email and password one more time.


#first need to give input

email=input("Enter the email: ")
password=input("Enter the password: ")

if email=='ashwini.nemlekar@gamilcom' and password=='1234':
  print ("login successful")
elif email=='ashwini.nemlekar@gamilcom' and password!='1234':
  print ('incorrect password')
  password=input("Enter password again: ")
  if password=='1234':
   print ('login sucessful')
  else:
   print('Pls try after 24hrs')
elif email!='ashwini.nemlekar@gamilcom' and password=='1234':
  print ('incorrect email')
  email=input('Enter email again')
  if email=='ashwini.nemlekar@gamilcom':
    print('login sucessful')
  else:
    print('Pls try after 24hrs')
else:
   print('email or password is incorrect')
```

```python
# Write a Python program to accept three numbers from the user.
# 1.Check which number is the minimum among the three numbers.
# 2.If all three numbers are the same, print "All three numbers are equal"


#1st i have to give input

a=int(input('Enter first number:'))
b=int(input('Enter second number:'))
c=int(input('Enter third number:'))

if a==b==c :
  print('All three numbers are equal')
elif a<b and a<c :
  print('minimum number is :',a)
elif b<c :
  print('minimum number is: ',b)
else:
  print('minimum number is: ',c)
```

```python
# Write a Python program to accept two numbers and a mathematical operator from the user.
# If the user enters any other operator or leaves it blank, print "Please use correct mathematical operation".


fnum=int(input("enter fnum: "))
snum=int(input("enter snum: "))

operator=input("enter operator: ")

if operator=="+" :
  print(fnum+snum)
elif operator=="-" :
  print(fnum-snum)
elif operator=="*" :
  print(fnum*snum)
elif operator=="/":
  print(fnum/snum)
else:
  print("Please use correct mathematical operation")

```
