```python
# Print the type of different values using type()
a = 10
b = 10.50
c = "hello"
d = True
e = [1,2,3,4]
f = (1,2,3,4)
g = {"Name" : "Ashwini", "Gender":"Female", "Age":30}


print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
```

```python
# Type conversion: int ↔ float ↔ str ↔ bool

# Q1. Convert int to float
a = 25
b = float(a)
print(b)
print(type(b))

# Q2. Convert float to int
c = 15.8
d = int(c)
print(d)
print(type(d))

# Q3. Convert int to string
e = 100
f = str(e)
print(f)
print(type(f))

# Q4. Convert string to int
g = "250"
h = int(g)
print(h)
print(type(h))

# Q5. Convert float to string
i = 45.5
j = str(i)
print(j)
print(type(j))

# Q6. Convert string to float
k = "25.75"
l = float(k)
print(l)
print(type(l))

# Q7. Convert int to bool
m = 1
n = bool(m)
print(n)
print(type(n))

# Q9. Convert bool to int
o = True
p = int(o)
print(p)
print(type(p))

# Q11. Convert string to bool
q = "Python"
r = bool(q)
print(r)
print(type(r))

# Q12. Convert empty string to bool
s = ""
t = bool(s)
print(t)
print(type(t))
```

```python
# String: reverse, count vowels, check palindrome, count words, change case

# Q1. Reverse a string
text = "Python"
result = text[::-1]
print(result)


# Q2. Reverse a string entered by the user
text = input("Enter a string: ")
result = text[::-1]
print(result)
```

```python
from itertools import count
# # Q3. Count the number of vowels in a string
str = "Hello Python"
count = 0
for i in str:
  if i.lower() in "aeiou":
    count = count + 1
print(count)


# # Q4. Count vowels in a string entered by the user

str = input("Enter the string")
count = 0
for i in str:
  if i.lower() in "aeiou":
    count = count + 1
print("Count of vowels in str: ",count)
```

```python
# # Q5. Check whether a string is a palindrome
text = "madam"
if text == text[::-1]:
  print("test is palindrome")
else:
  print("test is not palindrome")

# # Q6. Check whether a user-entered string is a palindrome
text = input("Enter a string: ")
if text == text[::-1]:
  print("test is palindrome")
else:
  print("test is not palindrome")
```

```python
# # Q7. Count the number of words in a string
text = "Python is very easy"
words = text.split()
count = len(words)
print(count)


# # Q8. Count words in a string entered by the user
text = input("Enter a sentence: ")
words = text.split()
count = len(words)
print("Count words in a string: ", count)
```
