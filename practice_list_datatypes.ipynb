```python
# # Q9. Find the smallest odd number without using min().
numbers = [12, 35, 18, 7, 42, 25, 9]
smallest = None
for i in numbers:
  if i % 2 != 0:
    if smallest is None or i < smallest:
      smallest = i
print(smallest)
```

```python
# # Q10. Find the difference between the largest and smallest number.
# # Do not use max() or min().
numbers = [20, 5, 45, 12, 67, 30]
largest = None
smallest = None

for i in numbers:
  if i % 2 == 0:
    if largest is None or i > largest:
      largest = i

  if i % 2 != 0:
    if smallest is None or i < smallest:
      smallest = i

difference = largest - smallest
print(difference)



```
