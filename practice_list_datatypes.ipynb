```python
# # Q20. Find the number closest to the average.

numbers = [10, 20, 30, 40, 50]
total = 0
count = 0
for i in numbers:
    total = total + i
    count = count + 1
avg = total / count
closest = numbers[0]
for i in numbers:
    if abs(i - avg) < abs(closest - avg):
        closest = i
print("Average:", avg)
print("Closest number:", closest)
```

```python
# Set: union, intersection, difference, and removing duplicates

# Q1. Create a set and print its elements.
numbers = {10, 20, 30, 40, 50}
print(numbers)


# Q2. Find the union of two sets.
# Union means all unique elements from both sets.
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A | B)


# Q3. Find the intersection of two sets.
# Intersection means common elements.
A = {10, 20, 30, 40}
B = {30, 40, 50, 60}
print(A & B)


# Q4. Find the difference A - B.
# Find elements present in A but not in B.
# Find elements present in B but not in A.
A = {1, 2, 3, 4, 5}
B = {3, 4, 5, 6, 7}
print(A - B)
print(B - A)


# Q5. Remove duplicates from a list using a set.
numbers = [10, 20, 10, 30, 20, 40, 30]
result = list(set(numbers))
print(result)

```

```python
# Dictionary: count the frequency of characters or words in a string

str = "hello"
frequency = {}
for i in str:
  if i in frequency:
    frequency[i] = frequency[i] + 1
  else:
    frequency[i] = 1
print(frequency)


```

```python
# Dictionary: store student records and find the topper

students = {
    "Ashwini": 85,
    "Nikhil": 90,
    "Neha": 95,
    "Amey": 80
}
topper = ""
highest = 0
for name in students:
    if students[name] > highest:
        highest = students[name]
        topper = name
print("Topper:", topper)
print("Highest Marks:", highest)
```
